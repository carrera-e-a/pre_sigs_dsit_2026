# Dataset Specifications
Develope a Dataset with the following characteristics:

- A Dataset with 300 servers

Columns Specification
- 1) **server_id**: Unique identifier (e.g.: SVR-001, SVR-002...)/
  2) **voltage**: Decimal number between 110 and 130 (uniform distribution).
  3)  **humidity**: Decimal number between 30 & 70 (uniform distribution).
  4)  **test_duration"**: Integer number between 50 & 150 (uniform distribution).

Failure Logic (**test_result**):
- If the **voltage > 125** and **humidity > 60** -> 80% Failure Probability (**test_result = 0**).
- In any other case -> 5% Falure Probability (**test_result = 0**)

(Remember: 0 = Fail, 1 = Pass)

```python
# Dataset Name: Generate Battery Data
# File Name: 02_Generate_Battery_Data.ipynb
# Folder: 06_Projects
# Creation Date: August 13th, 2026
# Creation Hour: 12:29 GTM-6
# Author: Edgar Carrera

import pandas as pd
import numpy as np

# seed generator
rng = np.random.default_rng(42)
# the difference with the legacy random.seed() on my perspective
# is that we can assign an object (rng) to the generator
# np.random.default_rng().

# server's number
n = 300

# column's specification
data = {
    'server_id': [f'SVR-{str(i).zfill(3)}' for i in range(1, n+1)],
    'voltage': rng.uniform(110, 130, size=n),
    'humidity': rng.uniform(low=30, high=70, size=n),
    'test_duration': rng.integers(50, 151, size=n)
    #in discrete uniform distributions, the high parameter needs
    # to be +1 of the upper limite we want to set
    # in this case: 150 + 1 = 151, so we can get number 150.
}

# We create a DataFrame (Pandas class) object called df, 
# which is formed by the data's dictionary values
df = pd.DataFrame(data)

# Method: Returns the fist 5 rows from the DataFrame (table)
# df object
# df.head()

# Attribute: Returns the number of rows and columns 
# from the Data Frame df object
# df.shape

## LOGIC OF TEST RESULTS

condition = (
    (df['voltage'] > 125) & 
    (df['humidity'] > 60)
    )

# Remember 0=FAIL, 1=PASS
df['test_result'] = np.where(
    condition,
    rng.choice([0,1], size=n, p=[0.8, 0.2]), # HIGH RISK
    rng.choice([0,1], size=n, p=[0.05, 0.95]) # LOW RISK
)

df.to_csv('battery_data.csv', index=False)

print("Successfully generated Dataset!!")
df.head()


    
```

# Probability Concept's Notes
The key difference between a **discrete uniform distribution** and a **continuous uniform distribution** is:

## Discrete (uniform) Distribution: 

**CONCEPT**: We assign probabilities to individual values.

P(X=5)>0

it can occur, like in a dice, it can occur that a favorable result is 5:


P(X=5) = (Number of Favorable results) / (number of possible results total)

P(X=5) = 1 / 6

So the probability of having a favorable result of 5 is 1/6.

The same would happen with the probability of having a favorable result (X) of 4 in a dice:

P(X=4) = 1/6

or a favorable result of 2:

P(X=2) = 1/6

This is due, that even though a dice has 6 possible values, it can only have 1 possible favorable result each time is thrown.

## Continuous (uniform) Distribution: 

**CONCEPT**: We assign probabilities to ranges, no to individual points.

P(X=5)=0

even when 5 is perfectly possible.

We instead use ranges to calculate. For example, if:

X∼U(0,6)

then

P(2≤X≤3)= (3−2) / (6-0) = 1 / 6

but 

P(X=2) = 0

And here appears a pretty important distinction which probably would help you a lot: *in a continuous (uniform) distribution, **what is constant in the uniform is not the probability of each point, but the probability density**. This is why the uniform graphic has an horizontal line.*

## How to diferentiate between a Function/Method and an Attribute in Python?

**The Key Rule (Piece of Cake):**

- IF it has a parenthesis () at the end: It is a **Function** or a **Method** (executes one action). E.g.: **df.head()**, **pd.DataFrame()**.
- IF it DOESN'T has a parenthesis () at the end: It is an **Attribute** (stores one value or state). E.g.: **df.shape**, **df.columns**.

The **parenthesis** are the ***inmediate visual key***. A **Method** or **Function** always is executed, that's why it has parenthesis (although sometimes it got arguments inside). An **Attribute** is just an internal variable, that is why is doesn't have parenthesis.
