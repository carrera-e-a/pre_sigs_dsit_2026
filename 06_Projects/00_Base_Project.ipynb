```python
# 00_Base_Project.ipynb

import pandas as pd

# Load the file you generated
df = pd.read_csv('manufacturing_data.csv')

print("=== FIRST 5 ROWS ===")
print(df.head())

print("\n=== GENERAL INFORMATION (Data Types) ===")
print(df.info())

print("\n=== DESCRIPTIVE STATISTICS (Numericals) ===")
print(df.describe())

# Business question 1: Which shift fails the most?
print("\n=== PASS RATE PER SHIFT ===")
print(df.groupby('shift')['test_result'].mean())

# Business question 2: Which station has the worst performance?
print("\n=== PASS RATE PER STATION ===")
print(df.groupby('station_id')['test_result'].mean().sort_values())
```

# Metrics Definitions

- **test_result = 1** means 'Pass' (Passed Test).
- **test result = 0** means 'Fail' (Failed Test).
- The **Failure Rate** is calculated as: 1 - mean(test_result).
- The **Pass Rate** is calculated as: mean(test_result).
