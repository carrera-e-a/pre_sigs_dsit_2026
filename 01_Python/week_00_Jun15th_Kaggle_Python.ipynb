# Excercise 7 - Progress percentage Function

For example, if you completed 2 from 10 modules, it must return 20%.

## Constraints
- Don't use fixed values inside the function.
- The function must serve for any course.
- Try the function with at least 3 different cases.

```python
# Excercise 7 - Progress Percentage Function
# Date: June 17th, 2026
# Hour: 23:37
# Author: E. Carrera

def Prog_Perc(subj, completed_mods, total_mods):
    x = (completed_mods / total_mods) * 100
    print(subj, "=", round(x,2),"%")
    
subject = ["Deutsch", "Spanish", "English", "Korean"]
mods_completed = [4, 3, 5, 9]
mods_total = [10, 9,6, 13]

for i in range(len(mods_completed)):
        Prog_Perc(subject[i], mods_completed[i],mods_total[i])
```

# Excercise 9 - Mini Automatic Report

Using variables, lists, functions and dictionaries, generate a small text report.

It must say something as:
Preparation report for SIGS
University ...
Weekly Planned Hours ...
Main Priorities ...
Estimated Progress ...

## Constraints
- The report must use saved data in variables/dictionaries
- It shall include at least one function.
- It shall include at least one list.
- It shall include at least one calculation.

```python
# Excercise #9 - Mini Automatic Report
# Date: June 18th, 2026
# Hour: 16:56
# Author: E. Carrera

sigs_dict = dict(name = "E. Carrera",
                 country = "Mexico",
                 university = "Tsinguha SIGS", 
                 program = "Data Science & Information Technology",
                 weekly_planned_hours = 16,
                 weekly_estimated_hours = 13.5,
                 main_priorities = ["Machine Learning", "Optimization", 
                                    "Decision Making"],
                 start_date = "September 1st, 2026",
                 end_date = "July 31st, 2028",
                 duration = 2
                )

#print(sigs_dict["name"])
                 
def progress_calc(planned_hours, estimated_hours):
    return round(((estimated_hours/planned_hours)*100),2)

def automatic_report(name, country, univ, program, week_plan, week_estim, start, end, dur, prog, main_prior):
    print("PREPARATION REPORT FOR TSINGHUA SIGS")
    print("")
    print("The student", name, "from", country, "has applied to the university", univ + ".")
    print("He has chosen the master's program of", program, "which has a duration of", dur, "years, starting from", start, "and finishing on", end + ".")
    print("The student", name, "has planned to study per week", week_plan, "hours, and had estimated in parallel a realistic", week_estim, "hours per week of study/work.")
    print("Thus, the student has compromised for a", str(prog) + "% progress every week.")
    print(name, "main priorities in terms of study would be:", main_prior, ".")    

automatic_report(sigs_dict["name"], sigs_dict["country"], 
                 sigs_dict["university"], sigs_dict["program"],
                 sigs_dict["weekly_planned_hours"],
                 sigs_dict["weekly_estimated_hours"],
                 sigs_dict["start_date"],
                 sigs_dict["end_date"],
                 sigs_dict["duration"],
                 progress_calc(sigs_dict["weekly_planned_hours"],
                 sigs_dict["weekly_estimated_hours"]),
                 sigs_dict["main_priorities"]
                )
                                  
    
```
